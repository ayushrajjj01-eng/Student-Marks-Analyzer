```python
import numpy as np

print("STUDENT MARKS ANALYZER")

# Generate marks for 50 students in 5 subjects
np.random.seed(42)

subjects = np.array([
    "Python",
    "Maths",
    "Statistics",
    "DBMS",
    "English"
])

marks = np.random.randint(0, 101, size=(50, 5))

print("\nStudent Marks:")
print(marks)

print("\nArray Shape:", marks.shape)
print("Dimensions:", marks.ndim)
print("Total Elements:", marks.size)
print("Data Type:", marks.dtype)
```

```python
# Calculate average marks of each student
student_avg = marks.mean(axis=1)

# Find the topper
topper = np.argmax(student_avg)

print("STUDENT PERFORMANCE ANALYSIS")
print("-" * 40)

print("Topper: Student", topper + 1)
print("Topper Average:", round(student_avg[topper], 2))

# Find subject-wise toppers
print("\nSUBJECT-WISE TOPPERS")

for i in range(len(subjects)):
    highest = np.argmax(marks[:, i])

    print(subjects[i], "- Student", highest + 1,
          "Marks:", marks[highest, i])

# Calculate subject-wise averages
print("\nSUBJECT-WISE AVERAGES")

for i in range(len(subjects)):
    print(subjects[i], ":", round(marks[:, i].mean(), 2))
```

```python
# Calculate grades for each student

def calculate_grade(avg):
    if avg >= 90:
        return "A+"
    elif avg >= 80:
        return "A"
    elif avg >= 70:
        return "B"
    elif avg >= 60:
        return "C"
    elif avg >= 50:
        return "D"
    else:
        return "F"

# Assign grades
grades = [calculate_grade(avg) for avg in student_avg]

# Display student grades
print("STUDENT GRADE REPORT")
print("-" * 40)

for i in range(len(student_avg)):
    print("Student", i + 1,
          "| Average:", round(student_avg[i], 2),
          "| Grade:", grades[i])
```

```python
# DATA VISUALIZATION

import matplotlib.pyplot as plt

# 1. Student-wise Average Marks
plt.figure(figsize=(10, 5))

plt.bar(range(1, len(student_avg) + 1), student_avg)

plt.title("Student-wise Average Marks")
plt.xlabel("Student Number")
plt.ylabel("Average Marks")
plt.xticks(range(1, len(student_avg) + 1))
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()


# 2. Subject-wise Average Marks
subject_avg = marks.mean(axis=0)

plt.figure(figsize=(10, 5))

plt.bar(subjects, subject_avg)

plt.title("Subject-wise Average Marks")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()


# 3. Student Performance Comparison
plt.figure(figsize=(10, 5))

plt.plot(range(1, len(student_avg) + 1),
         student_avg, marker='o')

plt.title("Student Performance Comparison")
plt.xlabel("Student Number")
plt.ylabel("Average Marks")
plt.grid(True)

plt.show()
```

```python
# FINAL STUDENT PERFORMANCE REPORT

print("=" * 50)
print("       STUDENT MARKS ANALYZER")
print("       FINAL PERFORMANCE REPORT")
print("=" * 50)

# Total number of students
total_students = len(student_avg)

# Calculate class average
class_average = student_avg.mean()

# Find topper
topper = np.argmax(student_avg)

# Count pass and fail students
pass_count = sum(avg >= 50 for avg in student_avg)
fail_count = total_students - pass_count

# Display report
print("\nTotal Students:", total_students)

print("Class Average:", round(class_average, 2))

print("Class Topper: Student", topper + 1)

print("Topper Average:", round(student_avg[topper], 2))

print("\nPASS STUDENTS:", pass_count)
print("FAIL STUDENTS:", fail_count)

print("\nSTUDENT-WISE RESULTS")
print("-" * 50)

for i in range(total_students):
    result = "PASS" if student_avg[i] >= 50 else "FAIL"

    print(
        "Student", i + 1,
        "| Average:", round(student_avg[i], 2),
        "| Grade:", grades[i],
        "| Result:", result
    )

print("\n" + "=" * 50)
print("       END OF PERFORMANCE REPORT")
print("=" * 50)
```
